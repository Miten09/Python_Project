```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

class SalesDataAnalyzer:

    def __init__(self):
        self.data = None
        self.figure = None

    def load_data(self, file_path):

        try:
            self.data = pd.read_csv(file_path)
            print("Dataset loaded successfully!")

        except FileNotFoundError:
            print("File not found!")

        except Exception as e:
            print("Error:", e)

    def explore_data(self, choice):

        if choice == 1:
            print(self.data.head())

        elif choice == 2:
            print(self.data.tail())

        elif choice == 3:
            print(self.data.columns)

        elif choice == 4:
            print(self.data.dtypes)

        elif choice == 5:
            self.data.info()

        else:
            print("Invalid choice!")

    def mathematical_operation(self, choice):

        if choice == 1:
            print(self.data["Product"])

        elif choice == 2:
            self.data["Bonus"] = self.data["Sales"] * 0.10
            print(self.data)

        elif choice == 3:
            print(self.data.sort_values(by="Sales"))

        elif choice == 4:
            print(self.data[self.data["Sales"] > 5000])

        elif choice == 5:
            self.data.drop_duplicates(inplace=True)
            print(self.data)

        else:
            print("Invalid choice!")

    def clean_data(self, choice):

        if choice == 1:

            missing_rows = self.data[
                self.data.isnull().any(axis=1)
            ]

            if missing_rows.empty:
                print("No missing values found!")

            else:
                print(missing_rows)

        elif choice == 2:

            numerical_columns = self.data.select_dtypes(
                include=np.number
            ).columns

            for col in numerical_columns:
                self.data[col] = self.data[col].fillna(
                    self.data[col].mean()
                )

            categorical_columns = self.data.select_dtypes(
                include="object"
            ).columns

            for col in categorical_columns:

                if self.data[col].isnull().any():

                    mode_value = self.data[col].mode()

                    if not mode_value.empty:
                        self.data[col] = self.data[col].fillna(
                            mode_value[0]
                        )

            print(self.data)
            print("\nRemaining Missing Values:")
            print(self.data.isnull().sum())

        elif choice == 3:

            self.data.dropna(axis=0, inplace=True)
            print(self.data)

        elif choice == 4:

            self.data["Payment_Mode"] = self.data[
                "Payment_Mode"
            ].fillna("UPI")

            self.data["Customer_Segment"] = self.data[
                "Customer_Segment"
            ].fillna("Consumer")

            self.data["Salesperson"] = self.data[
                "Salesperson"
            ].fillna("Mit")

            print(self.data)

        else:
            print("Invalid choice!")

    def descriptive_analysis(self, choice):

        if choice == 1:
            print(self.data.describe())

        elif choice == 2:
            print(self.data.describe(include="object"))

        elif choice == 3:
            print(self.data.describe(include="all"))

        elif choice == 4:
            print("Mean:")
            print(self.data.mean(numeric_only=True))

        elif choice == 5:
            print("Median:")
            print(self.data.median(numeric_only=True))

        elif choice == 6:
            print("Mode:")
            print(self.data.mode())

        elif choice == 7:
            print("Standard Deviation:")
            print(self.data.std(numeric_only=True))

        else:
            print("Invalid choice!")

    def visualize_data(self, choice):

        if choice == 1:

            x_axis = input("Enter x-axis column name: ")
            y_axis = input("Enter y-axis column name: ")

            data = self.data.dropna(subset=[x_axis, y_axis])

            data = data.groupby(x_axis)[y_axis].sum()

            plt.figure(figsize=(10, 6))
            self.figure = plt.gcf()

            plt.bar(
                data.index.astype(str),
                data.values,
                color='green'
            )

            plt.xlabel(x_axis)
            plt.ylabel(y_axis)
            plt.title(f"{y_axis} by {x_axis}")

            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        elif choice == 2:

            x_axis = input("Enter x-axis column name: ")
            y_axis = input("Enter y-axis column name: ")

            data = self.data.dropna(subset=[x_axis, y_axis])

            data = data.groupby(x_axis)[y_axis].sum()

            plt.figure(figsize=(10, 6))
            self.figure = plt.gcf()

            plt.plot(
                data.index.astype(str),
                data.values,
                color='blue',
                marker='o'
            )

            plt.xlabel(x_axis)
            plt.ylabel(y_axis)
            plt.title(f"{y_axis} Trend")

            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        elif choice == 3:

            x_axis = input("Enter x-axis column name: ")
            y_axis = input("Enter y-axis column name: ")

            data = self.data.dropna(subset=[x_axis, y_axis])

            data = data.groupby(x_axis)[y_axis].sum()

            plt.figure(figsize=(10, 6))
            self.figure = plt.gcf()

            plt.scatter(
                data.index.astype(str),
                data.values,
                color='red'
            )

            plt.xlabel(x_axis)
            plt.ylabel(y_axis)
            plt.title(f"{y_axis} vs {x_axis}")

            plt.tight_layout()
            plt.show()

        elif choice == 4:

            x_axis = input("Enter category column name: ")
            y_axis = input("Enter numerical column name: ")

            data = self.data.dropna(subset=[x_axis, y_axis])

            data = data.groupby(x_axis)[y_axis].sum()

            plt.figure(figsize=(8, 8))
            self.figure = plt.gcf()

            plt.pie(
                data.values,
                labels=data.index,
                autopct='%1.1f%%'
            )

            plt.title(f"{y_axis} Distribution")
            plt.show()

        elif choice == 5:

            y_axis = input("Enter numerical column name: ")

            plt.figure(figsize=(8, 6))
            self.figure = plt.gcf()

            plt.boxplot(
                self.data[y_axis].dropna()
            )

            plt.ylabel(y_axis)
            plt.title(f"Box Plot of {y_axis}")

            plt.show()

        elif choice == 6:

            y_axis = input("Enter numerical column name: ")

            plt.figure(figsize=(8, 6))
            self.figure = plt.gcf()
  
            plt.hist(
                self.data[y_axis].dropna(),
                bins=10,
                color='orange',
                edgecolor='black'
            )

            plt.xlabel(y_axis)
            plt.ylabel("Frequency")
            plt.title(f"Histogram of {y_axis}")

            plt.show()

        elif choice == 7:

            x_axis = input("Enter categorical column name: ")
            y_axis = input("Enter numerical column name: ")

            data = self.data.dropna(subset=[x_axis, y_axis])

            plt.figure(figsize=(10, 6))
            self.figure = plt.gcf()

            sns.violinplot(
                data=data,
                x=x_axis,
                y=y_axis,
                inner='quart'
            )

            plt.xlabel(x_axis)
            plt.ylabel(y_axis)
            plt.title(f"{y_axis} Distribution by {x_axis}")

            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        elif choice == 8:

            x_axis = input("Enter x-axis column name: ")

            print("Enter 3 numerical columns:")

            col1 = input("Enter first column: ")
            col2 = input("Enter second column: ")
            col3 = input("Enter third column: ")

            data = self.data.dropna(
                subset=[x_axis, col1, col2, col3]
            )

            plt.figure(figsize=(10, 6))
            self.figure = plt.gcf()

            plt.stackplot(
                range(len(data)),
                data[col1],
                data[col2],
                data[col3],
                labels=[col1, col2, col3]
            )

            plt.xticks(
                range(len(data)),
                data[x_axis],
                rotation=45
            )

            plt.xlabel(x_axis)
            plt.ylabel("Values")
            plt.title("Stack Plot")

            plt.legend()
            plt.tight_layout()
            plt.show()

        elif choice == 9:

            x_axis = input("Enter x-axis column name: ")
            y_axis = input("Enter y-axis column name: ")

            data = self.data.dropna(subset=[x_axis, y_axis])

            data = data.groupby(x_axis)[y_axis].sum()

            plt.figure(figsize=(10, 6))
            self.figure = plt.gcf()

            plt.step(
                data.index.astype(str),
                data.values,
                where='mid',
                color='purple'
            )

            plt.xlabel(x_axis)
            plt.ylabel(y_axis)
            plt.title(f"{y_axis} Step Chart")

            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        else:
            print("Invalid choice!")

    def save_visualization(self):

        if self.figure is None:
            print("Please generate a visualization first!")
            return

        file_name = input(
            "Enter file name (e.g. bar_plot.png): "
        )

        try:
            self.figure.savefig(file_name)
            print("Visualization saved successfully!")

        except Exception as e:
            print("Error saving visualization:", e)


s1 = SalesDataAnalyzer()


while True:


    print("Please select an option:")

    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:

        print("\n== Load Dataset ==")

        path = input(
            "Enter the path of the dataset (CSV file): "
        )

        s1.load_data(path)

    elif choice == 2:

        if s1.data is None:
            print("Please load dataset first!")

        else:

            print("\n== Explore Data ==")

            print("1. Display the first 5 rows")
            print("2. Display the last 5 rows")
            print("3. Display column names")
            print("4. Display data types")
            print("5. Display basic info")

            explore_choice = int(
                input("Enter your choice: ")
            )

            s1.explore_data(explore_choice)

    elif choice == 3:

        if s1.data is None:
            print("Please load dataset first!")

        else:

            print("\n== DataFrame Operations ==")

            print("1. Select Columns")
            print("2. Add New Column")
            print("3. Sort Data")
            print("4. Filter Data")
            print("5. Remove Duplicates")

            operation_choice = int(
                input("Enter your choice: ")
            )

            s1.mathematical_operation(operation_choice)

    elif choice == 4:

        if s1.data is None:
            print("Please load dataset first!")

        else:

            print("\n== Handle Missing Data ==")

            print("1. Display rows with missing values")
            print("2. Fill missing values with mean/mode")
            print("3. Drop rows with missing values")
            print("4. Replace missing values with specific values")

            missing_data_choice = int(
                input("Enter your choice: ")
            )

            s1.clean_data(missing_data_choice)

    elif choice == 5:

        if s1.data is None:
            print("Please load dataset first!")

        else:

            print("Generate Descriptive Statistics")

            print("1. Display numerical statistics")
            print("2. Display categorical statistics")
            print("3. Display complete statistics")
            print("4. Calculate Mean")
            print("5. Calculate Median")
            print("6. Calculate Mode")
            print("7. Calculate Standard Deviation")

            option = int(input("Enter your choice: "))

            s1.descriptive_analysis(option)

    elif choice == 6:

        if s1.data is None:
            print("Please load dataset first!")

        else:

            print(" Data Visualization ")

            print("1. Bar Plot")
            print("2. Line Plot")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Box Plot")
            print("6. Histogram")
            print("7. Violin Plot")
            print("8. Stack Plot")
            print("9. Step Chart")

            option = int(input("Enter your choice: "))

            s1.visualize_data(option)

    elif choice == 7:

        s1.save_visualization()

    elif choice == 8:

        print("Exiting the program. Goodbye!")
        break

    else:
        print("Invalid choice! Please select between 1 and 8.")
```
