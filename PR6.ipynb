```python

import numpy as np

print("Welcome to the numpy Analyzer!")

print("Choose an option: ")
print("1. Create a Numpy array")
print("2. Perform Mathematical Operations")
print("3. Combine or Split Arrays")
print("4. Search, Sort, or Filter arrays")
print("5. Compute Aggregates and Statistics")
print("6. Exit")

class DataAnalytics:
                
    def __init__(self,array):
        self.array = array
                    
    def display(self):
        return ("Array created successfully",self.array)


p1 = None

while True:

    choice = int(input("Enter your choice: "))

    if choice == 6:
        print("Thank you for using NumPy Analyzer!")
        break

    if choice in [2, 3, 4] and p1 is None:
        print("Please create an array first!")
        continue

    if choice == 1:
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")
        
        arrayChoice = int(input("Enter your choice: " ))
        
        if arrayChoice == 1:
            
            elements = input("Enter elements for the array seperated by space: ")
            final_list = np.array(list(map(int, elements.split())))
             
            p1 = DataAnalytics(final_list)
            print(p1.display())

            while True:

                print("1. Indexing")
                print("2. Slicing")
                print("3. Go back")

                operation_choice = int(input("Enter your choice: "))

                if operation_choice == 1:
                    index = int(input("Enter Index: "))
                    indexed_array = final_list[index]
                    print(indexed_array)

                if operation_choice == 2:
                    row_range = input("Enter the range (start:end): ")
                    start, end = map(int, row_range.split(":"))

                    sliced_array = final_list[start:end]
                    print(sliced_array)

                if operation_choice == 3:
                    break

        elif arrayChoice == 2:

            no_of_rows = int(input("Enter no of rows: "))
            no_of_columns = int(input("Enter no of columns: "))
            elements = input("Enter elements for the array seperated by space: ")
            final_list = np.array(list(map(int, elements.split())))

            if len(final_list) == no_of_rows * no_of_columns:
                final_array = final_list.reshape(no_of_rows,no_of_columns)
            else:
                print("Invalid number of elements!")
                continue
            
            p1 = DataAnalytics(final_array)
            print(p1.display())

            while True:

                print("1. Indexing")
                print("2. Slicing")
                print("3. Go back")

                operation_choice = int(input("Enter your choice: "))

                if operation_choice == 1:
                    row_index = int(input("Enter Row Index: "))
                    column_index = int(input("Enter Column Index: "))
                    indexed_array = final_array[row_index,column_index]
                    print(indexed_array)

                if operation_choice == 2:

                    row_range = input("Enter the row range (start:end): ")
                    column_range = input("Enter the column range (start:end): ")

                    row_start, row_end = map(int, row_range.split(":"))
                    col_start, col_end = map(int, column_range.split(":"))

                    sliced_array = final_array[row_start:row_end, col_start:col_end]

                    print(sliced_array)

                if operation_choice == 3:
                    break

        elif arrayChoice == 3:

            no_of_blocks = int(input("Enter no of blocks: "))
            no_of_rows = int(input("Enter no of rows: "))
            no_of_columns = int(input("Enter no of columns: "))
            elements = input("Enter elements for the array seperated by space: ")
            final_list = np.array(list(map(int, elements.split())))

            if len(final_list) == no_of_blocks * no_of_rows * no_of_columns:
                final_array = final_list.reshape(no_of_blocks,no_of_rows,no_of_columns)
            else:
                print("Invalid number of elements!")
                continue

            p1 = DataAnalytics(final_array)
            print(p1.display())

            while True:

                print("1. Indexing")
                print("2. Slicing")
                print("3. Go back")

                operation_choice = int(input("Enter your choice: "))

                if operation_choice == 1:
                    block_index = int(input("Enter Block Index: "))
                    row_index = int(input("Enter Row Index: "))
                    column_index = int(input("Enter Column Index: "))
                    indexed_array = final_array[block_index,row_index,column_index]
                    print(indexed_array)

                if operation_choice == 2:

                    block_range = input("Enter the block range (start:end): ")
                    row_range = input("Enter the row range (start:end): ")
                    column_range = input("Enter the column range (start:end): ")

                    block_start, block_end = map(int, block_range.split(":"))
                    row_start, row_end = map(int, row_range.split(":"))
                    col_start, col_end = map(int, column_range.split(":"))

                    sliced_array = final_array[block_start:block_end,row_start:row_end, col_start:col_end]

                    print(sliced_array)

                if operation_choice == 3:
                    break

    if choice == 2:
        print("1. Addition")
        print("2. Substraction")
        print("3. Multiplication")
        print("4. Divison")

        math_choice = int(input("Enter your choice: "))

        elements = input("Enter the same-size array elements: ")
        final_list = np.array(list(map(int, elements.split())))
        message, data = p1.display()

        if len(final_list) == data.size:

            final_list_reshape = final_list.reshape(data.shape)

            if math_choice == 1:
                print("Result of Addition:")
                print(data + final_list_reshape)

            if math_choice == 2:
                print("Result of Substraction:")
                print(data - final_list_reshape)

            if math_choice == 3:
                print("Result of Multiplication:")
                print(data * final_list_reshape)

            if math_choice == 4:
                print("Result of Division:")
                print(data / final_list_reshape)

        else:
            print("Both arrays must have same number of elements!")

    if choice == 3:
        print("1. Combine Arrays")
        print("2. Split Array")

        combine_choice = int(input("Enter your choice: "))

        message, data = p1.display()

        if combine_choice == 1:

            elements = input("Enter the elements of another array to combine: ")

            second_array = np.array(list(map(int, elements.split())))

            if len(second_array) == data.size:

                second_array = second_array.reshape(data.shape)

                print("Original Array:")
                print(data)

                print("Second Array:")
                print(second_array)

                combined_array = np.vstack((data, second_array))

                print("Combined Array (Vertical Stack):")
                print(combined_array)

            else:
                print("Both arrays must have same size!")

        elif combine_choice == 2:

            split_array = np.array_split(data, 2, axis=0)

            print("First Split Array:")
            print(split_array[0])

            print("Second Split Array:")
            print(split_array[1])

    if choice == 4:
        print("1. Search a value")
        print("2. Sort the array")
        print("3. Filter values")

        mychoice = int(input("Enter your choice: "))

        if mychoice == 1:
            value = int(input("Enter a value: "))
            message, data = p1.display()

            print("data",data)
            print("val",value)

            index = np.where(data == value)
            print(index)

        if mychoice == 2:
            message, data = p1.display()

            sorted_arr = np.sort(data)

            print("Sorted",sorted_arr)

        if mychoice == 3:
            message, data = p1.display()

            filter_value = int(input("Enter filter value: "))

            new_filter_array = data[data > filter_value]

            print("Filtered Array:")
            print(new_filter_array)

    if choice == 5:
        print("Aggregates and Statistics:")
        print("Choose an aggregate/statistical operation:")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")

        stats_choice = int(input("Enter your choice: "))

        message, data = p1.display()

        print("Original Array:")
        print(data)

        if stats_choice == 1:
            print("Sum of Array:", np.sum(data))

        elif stats_choice == 2:
            print("Mean of Array:", np.mean(data))

        elif stats_choice == 3:
            print("Median of Array:", np.median(data))

        elif stats_choice == 4:
            print("Standard Deviation of Array:", np.std(data))

        elif stats_choice == 5:
            print("Variance of Array:", np.var(data))

        else:
            print("Invalid choice!")

```
